```python
def f(sales):
    if sales <= 1:
        return 'Low'
    elif sales <= 5:
        return 'Medium'
    elif sales <= 10:
        return 'High'
    else:
        return 'Very High'
```

```python
df['Sales Category'] = df['Global'].apply(f)
```

```python
df
```

```python
def a(sales):
    if sales < 1:
        return 'Poor'
    elif sales < 5:
        return 'Average'
    elif sales < 10:
        return 'Good'
    else:
        return 'Excellent'
```

```python
df['Performance'] = df['Global'].apply(a)
```

```python
df
```

```python
def n(year):
    if year <= 2014:
        return 'Launch'
    elif year <= 2016:
        return 'Growth'
    elif year <= 2019:
        return 'stable'
    else:
        return 'Decline'
```

```python
df['Release Period'] = df['Year'].apply(n)
```
